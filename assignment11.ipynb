Let's build a random forest model to predict `MedianHouseValue`.

1. Create your model, set the engine to `ranger` with `importance = 'impurity'` and the mode as `regression`.
2. Fit the model using `MedianHouseValue` as the response variable and all other columns as predictors.

````r
set.seed(42)

housing_rf = rand_forest(trees = 500) |>
  set_engine("ranger", importance = "impurity",
             seed = 42, num.threads = 1) |>
  set_mode("regression") |>
  fit(MedianHouseValue ~ ., data = housing)

housing_rf
````

❓ What is the OOB error for this model?

**Answer:**

The OOB error is **0.23380**, reported as mean squared error (MSE) by `ranger` for regression. Each block group's OOB prediction uses trees that did not include it in their bootstrap samples. Taking the square root gives an OOB RMSE of **0.48352**, or about **$48,352** on the house-value scale. The MSE itself is in squared $100,000 units, not dollars. This is an internal OOB estimate, not a result from a separate test set.

````r
rf_engine = extract_fit_engine(housing_rf)
oob_mse = rf_engine$prediction.error

tibble(OOB_MSE = oob_mse, OOB_RMSE = sqrt(oob_mse))
````

Plot the feature importances using the `vip` package. Remember that you can use the `extract_fit_engine` function to get the `ranger` model object to send to the `vip` function.

Median income has the largest importance in this fitted model, and latitude and longitude are also useful predictors. These are impurity-based importance scores, not evidence of causal effects.

````r
vip(rf_engine, num_features = 8, geom = "col",
    aesthetics = list(fill = "steelblue")) +
  labs(title = "Random forest feature importance",
       subtitle = "Impurity-based importance",
       x = "Predictor", y = "Importance") +
  theme_minimal() +
  theme(plot.margin = margin(8, 24, 8, 8))
````
